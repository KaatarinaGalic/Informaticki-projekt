```python
#danceability vs energy
#Jesu li energicnije pjesme ujedno i plesnije?
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df_clean.sample(5000),  # uzorak radi brzine
    x='energy',
    y='danceability',
    alpha=0.5
)

plt.title("Odnos energije i plesnosti pjesama")
plt.xlabel("Energy")
plt.ylabel("Danceability")
plt.show()
```

X-os prikazuje razinu energy, odnosno koliko je pjesma energična (glasnoća, intenzitet, brzina).
Y-os prikazuje danceability, mjeru koliko je pjesma pogodna za ples (ritam, stabilnost tempa i struktura).

Graf prikazuje odnos između energije i plesnosti pjesama. Može se uočiti umjeren pozitivan odnos, što znači da energičnije pjesme u prosjeku imaju višu plesnost. Međutim, podaci su prilično raspršeni, što pokazuje da visoka energija ne garantira nužno visoku plesnost. Postoje i pjesme s visokom energijom, ali nižom plesnošću, kao i obrnuto. To ukazuje da su energija i plesnost povezane, ali da plesnost ovisi i o drugim glazbenim karakteristikama poput ritma, tempa i strukture pjesme.

Graf pokazuje pozitivnu korelaciju između energije i plesnosti, što znači da energičnije pjesme često imaju veći potencijal za ples.

```python
#Acousticness kroz godine, d li su moderne pjesme manje akusticne.?
acoustic_year = df_clean.groupby('year')['acousticness'].mean()

plt.figure(figsize=(12,6))
sns.lineplot(x=acoustic_year.index, y=acoustic_year.values)

plt.title("Prosječna akustičnost pjesama kroz godine")
plt.xlabel("Godina")
plt.ylabel("Acousticness")
plt.show()
```

Graf prikazuje promjenu prosječne akustičnosti pjesama kroz vrijeme. U ranijim godinama dominiraju pjesme s visokom akustičnošću, što je povezano s ograničenjima tehnologije snimanja i manjom prisutnošću elektroničkih instrumenata.

Od sredine 20. stoljeća nadalje vidljiv je kontinuirani pad akustičnosti, što ukazuje na sve veću upotrebu električnih i digitalnih instrumenata te produkcijskih tehnika. U novijem razdoblju vrijednosti akustičnosti stabiliziraju se na nižoj razini, što odražava suvremene trendove u glazbenoj produkciji.

```python
#popularnost kroz desetljeca
df_clean['decade'] = (df_clean['year'] // 10) * 10

plt.figure(figsize=(12,6))
sns.violinplot(
    data=df_clean,
    x='decade',
    y='popularity',
    density_norm='width'
)
plt.title('Distribucija popularnosti po desetljećima')
plt.xlabel('Desetljeće')
plt.ylabel('Popularnost')
plt.show()
```

Graf prikazuje distribuciju popularnosti pjesama po desetljećima. Uočljivo je da su pjesme iz ranijih desetljeća (1920.–1950.) imale vrlo nisku razinu popularnosti, što je djelomično posljedica ograničenih tehnoloških mogućnosti i kasnijeg nastanka streaming platformi.
Od 1960-ih godina nadalje vidljiv je postupni porast prosječne popularnosti, dok u novijim desetljećima (2000.–2020.) dolazi do značajnog povećanja popularnosti pjesama. Također, raspon popularnosti postaje širi, što ukazuje na veću raznolikost glazbenog sadržaja i veći broj iznimno popularnih pjesama u modernom razdoblju.

```python
#korelacijska heatmapa (međusobna povezanost audio značajki)
features = [
    'danceability','energy','valence',
    'acousticness','loudness',
    'speechiness','tempo','popularity'
]

corr = df_clean[features].corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Korelacijska matrica audio značajki')
plt.show()

```

Korelacijska matrica prikazuje stupanj povezanosti između odabranih audio značajki pjesama. Najizraženija pozitivna korelacija vidljiva je između varijabli energy i loudness, što je očekivano jer energičnije pjesme u pravilu imaju veću glasnoću.

Popularnost pjesama pokazuje umjerenu pozitivnu korelaciju s varijablama energy i loudness, dok je negativno povezana s varijablom acousticness. To sugerira da su energičnije i glasnije pjesme u prosjeku popularnije na Spotify platformi, dok su akustične pjesme manje zastupljene među najpopularnijima.
Ostale značajke, poput tempo i speechiness, ne pokazuju značajnu povezanost s popularnošću.

```python
#popularity vs danceability, postoji li trend između plesnosti i popularnosti
plt.figure(figsize=(8,6))
sns.regplot(
    data=df_clean,
    x='danceability',
    y='popularity',
    scatter_kws={'alpha':0.2}
)
plt.title('Danceability vs Popularity')
plt.xlabel('Danceability')
plt.ylabel('Popularity')
plt.show()
```

Graf prikazuje odnos između plesnosti (danceability) i popularnosti pjesama. Vidljiv je blagi uzlazni trend, što upućuje na umjerenu pozitivnu povezanost između ove dvije varijable.

Iako se pjesme svih razina popularnosti pojavljuju duž cijelog raspona plesnosti, popularnije pjesme češće se nalaze pri srednjim do višim vrijednostima plesnosti. To sugerira da plesnije pjesme u prosjeku ostvaruju veću popularnost na Spotify platformi, iako plesnost sama po sebi nije presudan faktor uspjeha.

```python
#energy vs loudness gdje je najveca koncetracija pjesama
plt.figure(figsize=(8,6))
plt.hexbin(
    df_clean['energy'],
    df_clean['loudness'],
    gridsize=50,
    cmap='viridis'
)
plt.colorbar(label='Gustoća')
plt.xlabel('Energy')
plt.ylabel('Loudness')
plt.title('Gustina pjesama: Energy vs Loudness')
plt.show()
```

Najveća gustoća pjesama nalazi se u području srednje do visoke energije (0.5–0.9) i loudnessa između −12 i −6 dB, što pokazuje da većina pjesama na Spotifyju ima dinamičan i glasan zvuk, prilagođen modernim produkcijskim standardima.
Ljubicasta boja oznacava malo ili nimalo pjesama dok zuta puno pjesama

```python
#kde-distibucija popularnosti
plt.figure(figsize=(8,5))
sns.kdeplot(
    df_clean['popularity'],
    fill=True
)
plt.title('Distribucija popularnosti pjesama')
plt.xlabel('Popularity')
plt.show()
```

```python
#pairplot multidimenzionalna analiza
sns.pairplot(
    df_clean[
        ['danceability','energy','valence','loudness','popularity']
    ],
    corner=True
)
plt.show()
```

Pairplot graf prikazuje međusobne odnose i distribucije odabranih audio značajki (danceability, energy, valence, loudness) te njihove povezanosti s popularnošću pjesama.

Vidljivo je da varijable energy i loudness pokazuju pozitivnu povezanost, dok je acousticness (nije prikazanaa) u prethodnim analizama pokazala negativnu povezanost s energijom i popularnošću.
Popularnost pjesama ne pokazuje snažnu linearnu povezanost s pojedinačnim audio značajkama, već je rezultat kombinacije više čimbenika, što potvrđuje složenost predviđanja glazbenog uspjeha.

```python
decades = sorted(df_clean['decade'].unique())
avg_pop = []

for d in decades:
    avg_pop.append(df_clean[df_clean['decade'] == d]['popularity'].mean())

plt.figure(figsize=(8,5))
plt.plot(decades, avg_pop, marker='o')
plt.title('Prosječna popularnost pjesama po desetljećima')
plt.xlabel('Desetljeće')
plt.ylabel('Prosječna popularnost')
plt.grid(True)
plt.show()
```

Graf prikazuje promjenu prosječne popularnosti pjesama po desetljećima. Uočen je jasan uzlazni trend, pri čemu novija desetljeća bilježe znatno višu prosječnu popularnost u odnosu na starija razdoblja.
Ovaj trend može se povezati s razvojem digitalnih tehnologija, dostupnošću glazbe putem streaming platformi te promjenama u načinu konzumacije glazbenog sadržaja. Najveći porast vidljiv je nakon 2000-ih godina, što se podudara s pojavom modernih streaming servisa.

```python
top_songs = (
    df_clean[['name', 'artists', 'popularity']]
    .sort_values(by='popularity', ascending=False)
    .head(10)
)

plt.figure(figsize=(8,5))
plt.barh(top_songs['name'], top_songs['popularity'])
plt.xlabel('Popularnost')
plt.title('Top 10 najpopularnijih pjesama')
plt.gca().invert_yaxis()
plt.show()
```

Graf prikazuje deset najpopularnijih pjesama u promatranom skupu podataka. Sve prikazane pjesme potječu iz novijeg razdoblja, što potvrđuje dominantnu ulogu suvremene glazbe na Spotify platformi.
Visoke vrijednosti popularnosti ovih pjesama ukazuju na snažan utjecaj globalnih trendova, izvođača s velikom bazom slušatelja te značaj marketinške prisutnosti u digitalnom okruženju.
